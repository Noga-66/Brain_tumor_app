```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Circle, Arc

# دالة رسم قلب
def draw_heart(ax, x, y, size):
    t = np.linspace(0, 2*np.pi, 200)
    X = size * 16 * np.sin(t)**3
    Y = size * (13*np.cos(t) - 5*np.cos(2*t) - 2*np.cos(3*t) - np.cos(4*t))
    ax.fill(X + x, Y + y, color='red')

fig, ax = plt.subplots()
ax.set_aspect("equal")
ax.set_xlim(0, 400)
ax.set_ylim(0, 400)
ax.axis("off")
#face
face = Circle((200, 200), 100, color="pink", ec='black')
ax.add_patch(face)
# eye
left_eye = Circle((160, 230), 15, color="black")
right_eye = Circle((240, 230), 15, color="black")
ax.add_patch(left_eye)
ax.add_patch(right_eye)
# smile
mouth = Arc((200,170), 100, 70, theta1=180, theta2=360, color="black", lw=3)
ax.add_patch(mouth)

#  القلب جنب الوش
draw_heart(ax, 300, 260, 0.8)

plt.title("Smile with Heart ")
plt.show()
```

```python
import matplotlib.pyplot as plt
def bresenham(x1, y1, x2, y2):
    dx = abs(x2 - x1)
    dy = abs(y2 - y1)
    sx = 1 if x2 > x1 else -1
    sy = 1 if y2 > y1 else -1

    p = 2 * dy - dx
    x, y = x1, y1
    points = [(x, y)]

    while x < x2:
        if p >= 0:
            y += sy
            p -= 2 * dx
        x += sx
        p += 2 * dy
        points.append((x, y))

    print("Plotted points:", points)
    plt.figure(figsize=(6, 6))
    for x,y in points:
        plt.scatter(x,y, color="red", s=100)
    plt.grid(True, linestyle="--", linewidth=0.5)
    plt.xlim(min(x1, x2) - 1, max(x1, x2) + 1)
    plt.ylim(min(y1, y2) - 1, max(y1, y2) + 1)
    plt.title("Bresenham Line")
    plt.show()

bresenham(5, 8, 9, 11)
```

```python
#Circle Using Midpoint Circle Algorithm (Matplotlib)
import matplotlib.pyplot as plt

def draw_circle(x_c, y_c, r):
    x = 0
    y = r
    p = 1 - r
    points = []

    def plot_points(x_c, y_c, x, y):
        points.extend([
            (x_c + x, y_c + y), (x_c - x, y_c + y),
            (x_c + x, y_c - y), (x_c - x, y_c - y),
            (x_c + y, y_c + x), (x_c - y, y_c + x),
            (x_c + y, y_c - x), (x_c - y, y_c - x)
        ])

    plot_points(x_c, y_c, x, y)

    while x < y:
        x += 1
        if p < 0:
            p += 2 * x + 1
        else:
            y -= 1
            p += 2 * x - 2 * y + 1
        plot_points(x_c, y_c, x, y)

    return points


circle_points = draw_circle(0, 0, 10)

x_coords, y_coords = zip(*circle_points)

plt.figure(figsize=(6, 6))
plt.scatter(x_coords, y_coords, color='blue', s=10)
plt.gca().set_aspect("equal", adjustable='box')
plt.title("Circle using Midpoint Circle Algorithm")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
# take an input from user
import matplotlib.pyplot as plt

def midpoint_circle(xc, yc, r):
    x = 0
    y = r
    p = 1 - r

    points = []

    def plot_symmetric_points(xc, yc, x, y):
        points.extend([
            (xc + x, yc + y), (xc - x, yc + y),
            (xc + x, yc - y), (xc - x, yc - y),
            (xc + y, yc + x), (xc - y, yc + x),
            (xc + y, yc - x), (xc - y, yc - x)
        ])

    # أول نقطة
    plot_symmetric_points(xc, yc, x, y)

    # رسم ربع الدائرة
    while x < y:
        x += 1
        if p < 0:
            p = p + 2 * x + 1
        else:
            y -= 1
            p = p + 2 * x - 2 * y + 1

        plot_symmetric_points(xc, yc, x, y)

    return points


# 🔹 إدخال المستخدم
xc = int(input("Enter center x: "))
yc = int(input("Enter center y: "))
r = int(input("Enter radius: "))

# 🔹 حساب النقاط
circle_points = midpoint_circle(xc, yc, r)

# 🔹 فصل الإحداثيات
x_coords, y_coords = zip(*circle_points)

# 🔹 الرسم
plt.figure(figsize=(6, 6))
plt.scatter(x_coords, y_coords)

plt.gca().set_aspect("equal", adjustable='box')
plt.title("Midpoint Circle Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)

plt.show()
```
